```python
import pandas as pd
import numpy as np

# Student Performance and Attendance Analysis
# Name: Aman Pal
# Admission No: 26B01010586

np.random.seed(42)

# Create dataset of 200 students
n = 200

data = {
    "Student_ID": range(1, n + 1),
    "Student_Name": [f"Student_{i}" for i in range(1, n + 1)],
    "Attendance": np.random.randint(50, 101, n),
    "Assignment_Marks": np.random.randint(30, 101, n),
    "Internal_Marks": np.random.randint(20, 101, n),
    "Final_Marks": np.random.randint(25, 101, n),
    "Study_Hours": np.random.randint(1, 9, n)
}

df = pd.DataFrame(data)

# Save raw dataset
df.to_csv("Raw_Student_Data.csv", index=False)

print("RAW DATASET")
print(df.head(10))
print("Total Records:", len(df))

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Check duplicates
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Remove duplicates
cleaned_df = df.drop_duplicates()

# Save cleaned dataset
cleaned_df.to_csv("Cleaned_Student_Data.csv", index=False)

# Display cleaned dataset
print("\nCLEANED DATASET")
print(cleaned_df.head(10))

print("\nFINAL OUTPUT")
print("Total Students:", len(cleaned_df))
print("Raw Dataset Saved Successfully!")
print("Cleaned Dataset Saved Successfully!")
print("Project Completed!")
```
